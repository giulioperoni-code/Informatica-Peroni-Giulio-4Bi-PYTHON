# Procedura di consegna di un esercizio

**Destinatario:** compagno di laboratorio

**Scopo:** spiegare come modificare il file, controllare la sintassi e consegnare il lavoro in modo ordinato.

## Passo 1 — Modifica del file

Questo passaggio esiste per aggiornare il contenuto della consegna e per evitare di lavorare su un file vecchio o incompleto. Se lo si salta, è facile consegnare materiale non corretto.

```python
numero_passi = 4
minuti_stimati = 20
file_in_commit = 3
print(numero_passi, minuti_stimati, file_in_commit)
```

## Passo 2 — Verifica della sintassi

Controllare la correttezza del documento prima della consegna riduce il rischio di errori visibili. Se lo si salta, il file può apparire incompleto o mal formattato.

## Passo 3 — Esecuzione locale

Eseguire il codice e controllare che non compaiano messaggi di errore è fondamentale per verificare la validità del lavoro. Se si salta questo passaggio, la consegna può contenere output errati.

```python
tempo_totale = numero_passi * minuti_stimati
print("Minuti totali:", tempo_totale)
```

## Passo 4 — Consegna

La consegna finale deve presentare una cartella ordinata, file verificati e un lavoro coerente con le istruzioni. Se lo si salta, la consegna può risultare incompleta o poco leggibile.

| Comando | Effetto |
| --- | --- |
| `git status` | mostra i file modificati |
| `python` | esegue script o notebook |
| `jupyter nbconvert --execute` | esegue il notebook localmente |

```python
numero_file = numero_passi + file_in_commit
print("File considerati:", numero_file)
```

> Prima della consegna, verificare sempre che il documento sia coerente, leggibile e privo di errori evidenti. Se il file non viene controllato, la consegna non è affidabile.
