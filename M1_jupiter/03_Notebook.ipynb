## Le celle di testo si scrivono in Markdown

Questa cella è una cella di testo e usa la sintassi studiata nelle prime otto ore del modulo: un
titolo con i cancelletti, l'*enfasi* con un asterisco, il **grassetto** con due, il codice in linea
come `print` fra virgolette inverse, un collegamento alla
[specifica CommonMark](https://spec.commonmark.org/) e una tabella.

| Tasti | Effetto |
|:---|:---|
| `Maiusc+Invio` | Esegue la cella e passa alla successiva |

Code, con la differenza che qui il sorgente e il risultato occupano lo stesso posto sulla pagina.

> Una cella di testo non descrive che cosa fa il codice sottostante, che si legge da sé,

I titoli delle celle di testo alimentano il sommario laterale di JupyterLab, che si apre dalla
barra a sinistra: sono la struttura del documento, non una scelta di dimensione del carattere.

## Il numero fra parentesi quadre

Il numero a sinistra di una cella di codice non è il numero della cella nella pagina: è il **numero
di esecuzione**, cioè la posizione di quella esecuzione nella sequenza delle richieste che il kernel
ha ricevuto dall'ultimo avvio. Una cella mai eseguita ha le parentesi vuote, `[ ]`, e una cella in
corso mostra `[*]`.

Esegui la cella qui sotto tre volte di seguito con `Ctrl+Invio`, che esegue senza spostare la
selezione, e guarda che cosa cambia a sinistra.

`````python
print("Cella eseguita: guarda il numero fra parentesi quadre qui a sinistra.")
`````

Il numero è cresciuto a ogni esecuzione mentre il testo stampato è rimasto lo stesso. Ne segue una
conseguenza che vale per tutto l'anno: guardando un notebook si può ricostruire l'ordine in cui il
suo autore ha eseguito le celle, e se i numeri non sono crescenti dall'alto verso il basso quel
notebook è stato eseguito fuori ordine.

## Guardare dentro la memoria del kernel

Il comando magico `%who` elenca i nomi definiti nel kernel. Non fa parte di Python: lo interpreta
il kernel, e per questo comincia con il segno di percentuale. È lo strumento diagnostico da usare
ogni volta che un notebook si comporta in modo inatteso, perché mostra che cosa esiste davvero
invece di che cosa è scritto nella pagina.

`````python
%who
`````

L'elenco contiene i nomi creati finora e nient'altro. Se hai eseguito le celle nell'ordine, sono
`ore_settimanali`, `presenti` e `settimane_del_modulo`.

La cella che segue usa due costrutti che si studieranno nel modulo M2, il ciclo `for` e la funzione
`range`, e li usa soltanto per produrre più righe di output con poche righe di codice. Non serve
capirne la sintassi adesso: basta osservare che il comando magico `%%time`, scritto sulla prima
riga, misura il tempo impiegato dall'intera cella e stampa il risultato dopo l'output normale.

`````python
%%time
for numero in range(1, 4):
    print("Riga numero", numero)
`````

## Che cosa resta in memoria dopo il riavvio del kernel

Il comando magico `%reset -f` svuota la memoria del kernel senza riavviarlo, e produce lo stesso
effetto di `Kernel > Restart` sulle variabili: dopo di esso nessuno dei nomi definiti finora
esiste più. Il `-f` evita la richiesta di conferma.

Esegui la cella che segue e poi quella successiva.

`````python
%reset -f
`````

`````python
%who
`````

Il kernel risponde `Interactive namespace is empty.`, cioè la memoria è vuota. Guarda però che cosa
è rimasto sulla pagina: tutti gli output prodotti prima sono ancora lì, sotto le loro celle, e i
numeri di esecuzione pure. Il notebook mostra `22` studenti presenti e una variabile `presenti` che
non esiste più da nessuna parte.

Questa discrepanza fra ciò che si vede e ciò che esiste è la sorgente di gran parte degli errori
che incontrerai quest'anno. Se ora eseguissi di nuovo, da sola, la cella che stampa `presenti`, il
kernel risponderebbe con un messaggio d'errore:

```text
NameError: name 'presenti' is not defined
```

Il codice della cella è corretto; è l'ordine di esecuzione a non esserlo. Provalo davvero, poi usa
`Kernel > Restart Kernel and Run All Cells` per rimettere il notebook in uno stato coerente.
