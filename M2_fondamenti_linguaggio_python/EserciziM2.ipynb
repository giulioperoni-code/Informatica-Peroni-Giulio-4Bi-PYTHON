## Esercizio 2

```python
# Input: ore, minuti e secondi
ore = int(input("Inserisci il numero di ore: "))
minuti = int(input("Inserisci il numero di minuti: "))
secondi = int(input("Inserisci il numero di secondi: "))

# Calcolo del totale in secondi
totale_secondi = ore * 3600 + minuti * 60 + secondi

# Stampa del risultato
print(f"Il totale in secondi è: {totale_secondi}")
```

# Esercizio 3

# Input: tempo totale e tempo trascorso
tempo_totale = int(input("Inserisci il tempo totale in secondi: "))
tempo_trascorso = int(input("Inserisci il tempo trascorso in secondi: "))

# Calcolo del tempo rimanente
tempo_rimanente = tempo_totale - tempo_trascorso

# Conversione in ore, minuti e secondi
ore = tempo_rimanente // 3600
minuti = (tempo_rimanente % 3600) // 60
secondi = tempo_rimanente % 60

# Stampa del risultato
print(f"Tempo rimanente: {ore} ore, {minuti} minuti e {secondi} secondi.")
