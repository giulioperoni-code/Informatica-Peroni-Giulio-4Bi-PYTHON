## Esercizio 2

```python
# Input dei dati
nome_studente = input("Inserisci il nome dello studente: ")
voto1 = float(input("Inserisci il primo voto: "))
voto2 = float(input("Inserisci il secondo voto: "))
voto3 = float(input("Inserisci il terzo voto: "))
ore_assenza = int(input("Inserisci le ore di assenza: "))

# Calcolo della media
media_voti = (voto1 + voto2 + voto3) / 3

# Determinazione del giudizio
if ore_assenza > 50:
    giudizio = "non valutabile"
elif media_voti >= 8:
    giudizio = "ottimo"
elif media_voti >= 6:
    giudizio = "sufficiente"
else:
    giudizio = "insufficiente"

# Stampa del risultato
print(f"{nome_studente}, media: {media_voti:.2f}, giudizio: {giudizio}")
```

# Esercizio 3

```python
# Legge in ingresso i valori iniziale, finale e passo
valore_iniziale = int(input("Inserisci il valore iniziale: "))
valore_finale = int(input("Inserisci il valore finale: "))
passo = int(input("Inserisci il passo: "))

# Controlla se il passo è positivo
if passo <= 0:
    print("Il passo deve essere un valore positivo.")
else:
    # Stampa l'intestazione della tabella
    print(f"{'Chilometri':>10} {'Miglia':>10} {'Metri':>10}")
    print("-" * 32)
    
    # Genera la tabella
    for km in range(valore_iniziale, valore_finale + 1, passo):
        miglia = km * 0.621371
        metri = km * 1000
        print(f"{km:>10} {miglia:>10.3f} {metri:>10}")
```

## Esercizio 5

```python
stringa: str = "Il laboratorio 4BI dura 22 ore"

vocali = 0
consonanti = 0
numeri = 0

for carattere in stringa.lower():
    if carattere.isdigit():

        numeri += 1
    if carattere in "aeiou":
        vocali += 1
    elif carattere.isalpha():
        consonanti += 1

print(
    f"Nella stringa ci sono {vocali} vocali, {consonanti} consonanti e {numeri} numeri."
)
```

## Esercizio 7

```python
# Legge in ingresso un intero non negativo
numero = int(input("Inserisci un numero intero non negativo: "))

# Inizializzazione delle variabili
somma_cifre = 0
numero_cifre = 0
cifra_maggiore = 0

# Ciclo while per calcolare somma delle cifre, numero di cifre e cifra maggiore
while numero > 0:

    cifra = numero % 10  # Estrae l'ultima cifra
    somma_cifre += cifra

    numero_cifre += 1
    if cifra > cifra_maggiore:
        cifra_maggiore = cifra
    numero //= 10  # Rimuove l'ultima cifra

# Stampa i risultati
print("Somma delle cifre:", somma_cifre)
print("Numero di cifre:", numero_cifre)
print("Cifra maggiore:", cifra_maggiore)
```

## Esercizio 8

```python
numero: int = 52
numero_binario: str = bin(numero)[2:]  # Converte il numero in binario e rimuove il prefisso '0b'

numero_convertire: bytes = 10101000101001000101012

x = numero
x_bin = ''
while x > 0:
    x_bin = str(x % 2) + x_bin
    x //= 2
if x_bin == '':
    x_bin = '0'
    
confronto = "uguale" if numero_binario == x_bin else "diverso"

numero_convertire_decimale = 0
potenza = 0
inputcorretto:bool = True
for cifra in (str(numero_convertire)):
    if cifra not in "01":
        inputcorretto = False
        break
    numero_convertire_decimale += int(cifra) * (2 ** potenza)
    potenza += 1


print(f"Il numero {numero} in binario è {numero_binario}.")
print(confronto)
if inputcorretto:
    print(numero_convertire_decimale)
else:
    print("input errato")
```
